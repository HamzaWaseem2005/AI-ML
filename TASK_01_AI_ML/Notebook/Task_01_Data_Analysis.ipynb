PART 3 – EXPLORATORY DATA ANALYSIS (EDA)

REPORT FOR THE  SUMMARY IS:

 3.2 MeanFormula: $\text{Mean} = \frac{\Sigma x}{n}$Explanation: When you take the average for things like study hours and exam scores, you get a solid baseline of what a typical student in this dataset looks like, though it can sometimes get slightly pulled up or down if there are a few extreme slackers or overachievers.

 3.3 MedianExplanation: The median helps ground things by showing the exact middle score when sorted from lowest to highest. It is particularly useful because it proves how the typical student performed without letting extreme cases or outliers skew the picture.

 3.4 ModeExplanation: The mode identifies the most frequently occurring value. While numerical columns might not show a clear peak, it is exceptionally helpful for categorical variables to highlight the most common profile among the students.

 3.5 Standard Deviation
Explanation: Standard deviation measures how much student scores vary from the average. A higher spread indicates a wide performance gap across the cohort, showing that student outcomes stretch across a broad spectrum depending on their study habits.

 Descriptive Statistics

```python
numerical_df = df.select_dtypes(include=['number'])
summary={
    'Mean': numerical_df.mean(),
    'Median': numerical_df.median(),
    'Minimum': numerical_df.min(),
    'Maximum': numerical_df.max(),
    'Range': numerical_df.max() - numerical_df.min(),
    'Std Dev': numerical_df.std(),
    'Variance': numerical_df.var()
}
print(summary)
```

Correlation Analysis

PART 4 – DATA VISUALIZATION

1.Histogram

What the graph represents: This histogram displays the distribution and frequency of student Exam_Score values across the entire dataset.

What pattern is visible: The scores form a roughly bell-shaped curve, clustering heavily around the middle grades, with fewer students scoring extremely high or extremely low.

What the result means: A majority of students perform at an average level, showing standard academic achievement rather than heavy skewing toward failure or perfection

```python


plt.figure(figsize=(8, 5))
sns.histplot(df['Exam_Score'], bins=20, kde=True)
plt.title('Distribution of Student Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Count')
plt.show()
```

2.Bar Plot

What the graph represents: This bar chart compares the average Exam_Score across different categories of Parental_Involvement (such as Low, Medium, and High).

What pattern is visible: The height of the bars increases steadily from low involvement to high involvement, showing a clear step-up in performance as family support rises.

What the result means: Students who receive higher levels of parental guidance and academic backing consistently achieve higher average exam scores than peers with minimal support.

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Parental_Involvement', y='Exam_Score', data=df)
plt.title('Average Exam Score by Parental Involvement')
plt.xlabel('Parental Involvement')
plt.ylabel('Average Exam Score')
plt.show()
```

3.Pie Chart

What the graph represents: This pie chart shows the proportional breakdown of students categorized by their level of Access_To_Resources (such as Low, Medium, or High).

What pattern is visible: The chart divides the student cohort into distinct percentage slices, instantly highlighting which resource tier forms the largest portion of the population.

What the result means: It reveals whether the majority of students enjoy adequate academic resources or if a significant fraction operates under restricted access conditions.

```python
plt.figure(figsize=(7, 7))
category_counts = df['Access_to_Resources'].value_counts()
plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%')
plt.title('Student Access to Educational Resources')
plt.show()
```

4.Scatter Plot

What the graph represents: This scatter plot maps individual student data points to show the direct relationship between Hours_Studied on the x-axis and Exam_Score on the y-axis.

What pattern is visible: The data points trend upward from bottom-left to top-right, showing that students who spend more hours studying cluster in higher grade zones.

What the result means: It visually confirms a positive association between study time and academic performance, aligning with our earlier correlation matrix findings.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Hours_Studied', y='Exam_Score', data=df)
plt.title('Relationship Between Hours Studied and Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()
```

5.Line Chart

What the graph represents: This line chart traces the trajectory of average Exam_Score mapped sequentially across different increments of Hours_Studied.

What pattern is visible: The line rises steadily from left to right, illustrating a smooth, progressive upward trajectory as study hours increase.

What the result means: It highlights a consistent linear progression where each additional tier of study time yields predictable gains in average performance.

```python
plt.figure(figsize=(8, 5))
trend_data = df.groupby('Hours_Studied')['Exam_Score'].mean().reset_index()
sns.lineplot(x='Hours_Studied', y='Exam_Score', data=trend_data)
plt.title('Trend of Average Exam Score ')
plt.xlabel('Hours Studied')
plt.ylabel('Average Exam Score')
plt.show()
```

6.Box Plot

What the graph represents: This box plot summarizes the statistical distribution of student Exam_Score values, highlighting the median, quartiles, and potential outliers.

What pattern is visible: The central box displays the interquartile range (IQR) containing the middle 50% of scores, with a clear median line running through it, while the whiskers extend to show the overall spread.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y=df['Exam_Score'])
plt.title('Box Plot of Student Exam Scores ')
plt.ylabel('Exam Score')
plt.show()
```

7.Correlation Heatmap

What the graph represents: This heatmap visualizes the correlation matrix of all numerical variables in the dataset, using a color-coded grid to represent the strength and direction of linear relationships.

What pattern is visible: Darker warm tones highlight strong positive relationships (such as between Attendance, Hours_Studied, and Exam_Score), while neutral tones dominate pairings with near-zero correlation like sleep or physical activity.

```python
plt.figure(figsize=(9, 6))
numeric_df = df.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()
sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Heatmap ')
plt.show()
```

8.Count Plot

What the graph represents: This count plot displays the total frequency (number of students) falling into discrete performance categories (Low, Medium, High) derived from their exam scores.

```python
plt.figure(figsize=(8, 5))
df['Performance_Tier'] = pd.cut(df['Exam_Score'], bins=3, labels=['Low', 'Medium', 'High'])
sns.countplot(x='Performance_Tier', data=df)
plt.title('Count of Students ')
plt.xlabel('Performance Tier')
plt.ylabel('Number of Students')
plt.show()
```

9.Pair Plot

What the graph represents: This pair plot creates a grid of scatter plots and kernel density estimations (KDE), showing pairwise relationships across all combinations of key numerical variables simultaneously.

What pattern is visible: The diagonal displays distribution curves for each individual feature, while the off-diagonal panels map bivariate interactions, cleanly showing linear trends for variables like Attendance and Exam_Score.

```python
plt.figure(figsize=(10, 10))
key_columns = ['Hours_Studied', 'Attendance', 'Previous_Scores', 'Exam_Score']
sns.pairplot(df[key_columns], diag_kind='kde')
plt.suptitle('Pair Plot ', y=1.02)
plt.show()
```

10. Violin Plot

What the graph represents: This violin plot combines a box plot with a kernel density estimation (KDE) curve to display the full distribution density of Exam_Score across different categories of Tutoring_Sessions.

What pattern is visible: The symmetrical "violins" widen at points where student scores are most densely concentrated, showing both the median and the overall spread for each tutoring tier.

What the result means: It reveals how student performance is distributed around the average within each tutoring bracket, showing whether scores cluster tightly or vary widely.

```python
plt.figure(figsize=(8, 5))
sns.violinplot(x='Tutoring_Sessions', y='Exam_Score', data=df)
plt.title('Violin Plot of Exam Scores ')
plt.xlabel('Tutoring Sessions')
plt.ylabel('Exam Score')
plt.show()
```
